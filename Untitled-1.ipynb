```python
#פונקציה לכותרות
def get_headers(df):
    """
    הפונקציה מקבלת טבלה (DataFrame) 
    ומדפיסה את כל שמות העמודות שלה כשורה נפרדת
    """
    headers = df.columns.tolist()
    print(f"סה'כ עמודות: {len(headers)}\n")
    for col in headers:
        print(col)


def search_in_column(df, column_name, search_value):
    """
    הפונקציה מקבלת טבלה, שם עמודה, וערך לחיפוש,
    ומחזירה את כל השורות התואמות
    """
    # בדיקה ששם העמודה אכן קיים בקובץ כדי למנוע שגיאות
    if column_name not in df.columns:
        print(f"שגיאה: העמודה '{column_name}' לא קיימת בטבלה זו.")
        return None
    
    # ביצוע החיפוש
    result = df[df[column_name] == search_value]
    
    # הדפסת הודעה קצרה על כמות התוצאות שנמצאו
    print(f"נמצאו {len(result)} שורות מתאימות:")
    
    # החזרת התוצאה
    return result        
```

```python
get_headers(TALMIDIM_AM_2026)
my_result = search_in_column(TALMIDIM_AM_2026, 'MISPAR_ZEHUT', 331259812)
display(my_result)
```

```python
#פונקציה לחיפוש לפי ת.ז אם מופיע כמה פעמים ואם יש סטטוס שונה

def get_id_data_TALMIDIM_AM( id_number):
    # 1. מגדירים את שתי העמודות שרוצים להשאיר (שנה את השמות לשמות האמיתיים אצלך)
    columns_to_keep = ['MISPAR_ZEHUT', 'CODE_STATUS_RISHUM_TA']
    
    # 2. מסננים לפי תעודת הזהות (שנה את 'ID' לשם עמודת תעודת הזהות אצלך)
    filtered_df = TALMIDIM_AM_2026[TALMIDIM_AM_2026['MISPAR_ZEHUT'] == id_number]
    
    # 3. מחזירים רק את שתי העמודות המבוקשות
    return filtered_df[columns_to_keep]

#פונקציה להדפסה של השורות ועמודות
def show_df_columns(df):
    # רצים בלולאה ישירות על העמודות של ה-DataFrame שהתקבל
    for index, column_name in enumerate(df.columns):
        print(f"{index}: {column_name}")

```
